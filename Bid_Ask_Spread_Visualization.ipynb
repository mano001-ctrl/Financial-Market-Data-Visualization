```python
import numpy as np
import matplotlib.pyplot as plt

# Generate random bid and ask orders
np.random.seed(42)
bid_prices = np.random.uniform(99.0, 99.8, 10)
ask_prices = np.random.uniform(100.2, 101.0, 10)
bid_sizes = np.random.randint(1, 10, 10)
ask_sizes = np.random.randint(1, 10, 10)

# Determine the best bid and ask
best_bid = max(bid_prices)
best_ask = min(ask_prices)

# Create the plot
fig, ax = plt.subplots(figsize=(8, 5))
ax.set_xlim(98.8, 101.2)
ax.set_ylim(0, max(max(bid_sizes), max(ask_sizes)) + 2)
ax.set_xlabel("Price")
ax.set_ylabel("Order Size")
ax.set_title("Bid-Ask Spread Visualization")

# Plot bid and ask orders
ax.scatter(bid_prices, bid_sizes, color='blue', label='Bid Orders', alpha=0.7)
ax.scatter(ask_prices, ask_sizes, color='red', label='Ask Orders', alpha=0.7)

# Highlight the best bid and ask
ax.axvline(best_bid, color='blue', linestyle='--', label='Best Bid')
ax.axvline(best_ask, color='red', linestyle='--', label='Best Ask')

# Highlight the bid-ask spread
ax.plot([best_bid, best_ask], [max(bid_sizes) + 1, max(bid_sizes) + 1], 'g-', lw=2, label='Bid-Ask Spread')

# Add legend
ax.legend()

# Show the plot
plt.show()
```

Here is a static visualization of the bid-ask spread:

Blue dots represent bid orders.
Red dots represent ask orders.
Dashed blue and red lines indicate the best bid and best ask, respectively.
Green line represents the spread between the highest bid and the lowest ask.
