```python
import numpy as np
import matplotlib.pyplot as plt

# Initialize order book levels
order_book = {
    'bid_prices': [99.5, 99.6, 99.7, 99.8, 99.9],
    'bid_sizes': [20, 25, 30, 35, 40],
    'ask_prices': [100.1, 100.2, 100.3, 100.4, 100.5],
    'ask_sizes': [20, 25, 30, 35, 40]
}

# Adding a spoof order
spoof_bid_prices = order_book['bid_prices'][:]
spoof_bid_sizes = order_book['bid_sizes'][:]
spoof_bid_sizes[-1] = 90  # Large spoof order at highest bid price

# Simulated execution on the ask side
executed_ask_sizes = order_book['ask_sizes'][:]
executed_ask_sizes[0] -= 15  # Trade execution

# Creating the plot
fig, ax = plt.subplots(figsize=(8, 6))
bar_width = 0.2

# Plot bid orders (including spoof order)
ax.bar(spoof_bid_prices, spoof_bid_sizes, width=bar_width, color='blue', alpha=0.7, label='Bids (Spoofing)')

# Plot ask orders after execution
ax.bar(order_book['ask_prices'], executed_ask_sizes, width=bar_width, color='red', alpha=0.7, label='Asks (Trade Executed)')

# Labels and titles
ax.set_title("Order Spoofing in Trading")
ax.set_xlabel("Price")
ax.set_ylabel("Order Size")
ax.set_xlim(99.4, 100.6)
ax.set_ylim(0, 100)
ax.legend()

# Show the plot
plt.show()
```

Order Spoofing in Trading
What is Order Spoofing?
Order spoofing is a market manipulation tactic where a trader places large buy or sell orders to create a false impression of demand or supply. The goal is to deceive other market participants into reacting to these fake orders, after which the spoofer cancels them once their real trades are executed at favorable prices.

How Spoofing Works (As Shown in the Visual)
Normal Market Orders

The order book initially consists of normal bid (buy) and ask (sell) orders.
Market participants place orders based on supply and demand.
Spoof Order Placement

A spoofer places a large, deceptive buy order at the highest bid price (blue bar).
This artificially inflates demand, making it seem like buying pressure is increasing.
Other traders react by raising their bid prices or executing sell orders prematurely.
Trade Execution

The spoofer executes a real buy order on the ask side (red bars).
The presence of fake demand tricks sellers into offering lower prices.
Spoof Order Cancellation

Once the real trade is complete, the large bid order is canceled.
The artificial demand vanishes, often causing prices to fall back.
Impact of Spoofing
Misleads other traders into making suboptimal decisions.
Creates artificial volatility that disrupts price discovery.
Regulated as illegal by financial authorities like the SEC and CFTC.
This visualization demonstrates how spoofing manipulates the market and why regulatory agencies actively monitor for such practices.
