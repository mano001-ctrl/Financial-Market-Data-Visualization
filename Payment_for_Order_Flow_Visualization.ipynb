```python
import matplotlib.pyplot as plt
import networkx as nx

# Create a directed graph
G = nx.DiGraph()

# Define nodes
nodes = ["Retail Investor", "Broker", "Market Maker", "Stock Exchange"]

# Add nodes
G.add_nodes_from(nodes)

# Define edges (Order Flow)
edges = [
    ("Retail Investor", "Broker", "Order Execution"),
    ("Broker", "Market Maker", "Payment for Order Flow"),
    ("Market Maker", "Stock Exchange", "Price Discovery"),
    ("Market Maker", "Retail Investor", "Trade Execution"),
]

# Add edges
for edge in edges:
    G.add_edge(edge[0], edge[1], label=edge[2])

# Positioning the nodes
pos = {
    "Retail Investor": (0, 2),
    "Broker": (2, 2),
    "Market Maker": (4, 2),
    "Stock Exchange": (6, 2),
}

# Increase the figure size for better readability
plt.figure(figsize=(15, 7))

# Draw the graph with larger nodes and font sizes
nx.draw(G, pos, with_labels=True, node_color="lightblue", edge_color="black",
        node_size=5000, font_size=12, font_weight="bold", arrows=True)

# Add edge labels with larger font size
edge_labels = {(edge[0], edge[1]): edge[2] for edge in edges}
nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels, font_size=11)

# Show plot with updated title
plt.title("Payment for Order Flow (PFOF) Process", fontsize=16)
plt.show()
```
