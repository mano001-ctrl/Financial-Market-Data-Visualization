```python
import pandas as pd
import numpy as np
import plotly.express as px
from datetime import datetime, timedelta

# Simulate S&P 500 companies data
num_companies = 100

# Generate random data for company headquarters within the US
data = {
    "company": [f"Company {i}" for i in range(1, num_companies + 1)],
    "latitude": np.random.uniform(25, 49, num_companies),  # Random latitudes in the U.S.
    "longitude": np.random.uniform(-125, -66, num_companies),  # Random longitudes in the U.S.
    "earnings": np.random.uniform(1e7, 1e10, num_companies),  # Earnings between $10M and $10B
    "date": [datetime(2020, 1, 1) + timedelta(days=90 * int(i)) for i in np.random.randint(0, 50, num_companies)],  # Random quarters
}

# Create a DataFrame
df = pd.DataFrame(data)

# Format date for visualization
df['date'] = df['date'].dt.strftime('%Y-%m-%d')

# Display the first few rows of the simulated data
print(df.head())
```

```python
# Create an interactive map to visualize earnings
fig = px.scatter_mapbox(
    df,
    lat="latitude",
    lon="longitude",
    size="earnings",  # Marker size based on earnings
    color="earnings",  # Color by earnings
    hover_name="company",  # Show company name on hover
    hover_data={"earnings": ":,.2f", "date": True},  # Display earnings with formatting
    animation_frame="date",  # Add a time slider for quarterly earnings
    title="Simulated Earnings of S&P 500 Companies by Headquarters",
    mapbox_style="carto-positron",
    size_max=20,
    zoom=3,  # Adjust zoom to focus on the U.S.
)

# Display the map
fig.show()
```
