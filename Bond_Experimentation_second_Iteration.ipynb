```python
import matplotlib.pyplot as plt

# Data for visualization
categories = ['Coupon Rate (%)', 'Liquidity (1-10)', 'Risk Level (1-10)']
fgn_values = [14.8, 6, 8]  # Example values for FGN
r186_values = [10.5, 9, 6]  # Example values for R186

# Create the bar chart
x = range(len(categories))
width = 0.35  # Width of bars

plt.bar([p - width/2 for p in x], fgn_values, width, label='FGN 14.8% Apr 2049')
plt.bar([p + width/2 for p in x], r186_values, width, label='R186')

# Add labels and title
plt.xticks(x, categories)
plt.ylabel('Values')
plt.title('Comparison of FGN 14.8% Bond and R186 Bond')
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Data for the radar chart
categories = ['Coupon Rate', 'Liquidity', 'Risk']
fgn_values = [14.8, 6, 8]
r186_values = [10.5, 9, 6]

# Radar chart setup
angles = np.linspace(0, 2 * np.pi, len(categories), endpoint=False).tolist()
fgn_values += fgn_values[:1]
r186_values += r186_values[:1]
angles += angles[:1]

# Plot
fig, ax = plt.subplots(figsize=(6, 6), subplot_kw={'polar': True})
ax.plot(angles, fgn_values, label='FGN 14.8% Apr 2049')
ax.fill(angles, fgn_values, alpha=0.25)
ax.plot(angles, r186_values, label='R186')
ax.fill(angles, r186_values, alpha=0.25)

ax.set_yticks([5, 10, 15])
ax.set_xticks(angles[:-1])
ax.set_xticklabels(categories)

plt.legend()
plt.title('Comparison of FGN and R186 Bonds')
plt.show()
```
