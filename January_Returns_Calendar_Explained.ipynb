
# Daily Returns and Point Movements

This notebook explains step-by-step how to calculate and visualize daily returns and point movements for the month of January in a calendar format. 

The process includes downloading financial data, calculating returns and movements, organizing the data into a calendar-style structure, and creating a heatmap for better visualization.


```python

import yfinance as yf
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import calendar
```

```python

# Define the ticker for the stock (example: Apple)
tickers = ['AAPL']

# Download daily data from Yahoo Finance
data = yf.download(tickers, start="2023-01-01", end="2023-02-01", interval="1d")['Adj Close']

# Calculate daily returns (percentage change)
daily_returns = data.pct_change().dropna() * 100

# Calculate daily point movements
daily_point_movement = data.diff().dropna()
```

```python

# Filter data for the month of January
january_returns = daily_returns[daily_returns.index.month == 1]
january_point_movements = daily_point_movement[daily_point_movement.index.month == 1]

# Calculate the daily average return for January
january_daily_avg_return = january_returns.mean(axis=1)
```

```python

# Create a calendar-style DataFrame for visualization
calendar_data = january_daily_avg_return.reset_index()
calendar_data['Year'] = calendar_data['Date'].dt.year
calendar_data['Day'] = calendar_data['Date'].dt.day
calendar_data['Weekday'] = calendar_data['Date'].dt.weekday
calendar_data['Week'] = calendar_data['Date'].dt.isocalendar().week

# Add point movements to the calendar data
calendar_data['Point Movement'] = january_point_movements.values
calendar_data['Combined'] = calendar_data[0].round(2).astype(str) + "% (" + calendar_data['Point Movement'].round(2).astype(str) + ")"
```

```python

# Pivot data for calendar structure
calendar_pivot = calendar_data.pivot(index='Week', columns='Weekday', values='Combined')

# Adjust column names for weekdays
weekday_labels = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
calendar_pivot.columns = weekday_labels[:len(calendar_pivot.columns)]

# Plot the calendar heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(calendar_pivot, annot=True, fmt="s", cmap="coolwarm", cbar=False, linewidths=0.5)

plt.title("January Daily Returns and Point Movements in Calendar Format", fontsize=16)
plt.xlabel("Day of Week", fontsize=12)
plt.ylabel("Week", fontsize=12)
plt.tight_layout()
plt.show()
```
