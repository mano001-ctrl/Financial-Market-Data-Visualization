```python
from pytrends.request import TrendReq
import pandas as pd
import plotly.graph_objects as go

# Initialize pytrends
pytrends = TrendReq(hl='en-US', tz=360)

# Define keywords/themes of interest (South African banks)
keywords = [
    "Green Finance", "Sustainable Investing", "Climate Risk", "Carbon Tax"
]

# Build payload
pytrends.build_payload(keywords, cat=0, timeframe='today 12-m', geo='ZA', gprop='')

# Fetch interest over time data
interest_over_time_df = pytrends.interest_over_time()

# Remove the 'isPartial' column if present
if 'isPartial' in interest_over_time_df.columns:
    interest_over_time_df = interest_over_time_df.drop(columns=['isPartial'])

# Create a Plotly figure
fig = go.Figure()

# Add a line for each keyword
for keyword in keywords:
    fig.add_trace(go.Scatter(
        x=interest_over_time_df.index,
        y=interest_over_time_df[keyword],
        mode='lines',
        name=keyword
    ))

# Add layout details
fig.update_layout(
    title="Google Trends: Interest Over Time for South African Banks",
    xaxis_title="Date",
    yaxis_title="Search Interest",
    legend_title="Banks",
    template="plotly_white",
    width=1000,
    height=600
)

# Show the interactive chart
fig.show()
```
